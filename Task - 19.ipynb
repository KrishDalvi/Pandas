```python
# Question 4

import pandas as pd
import numpy as np

# 1. Load the dataset
df = pd.read_csv('amazon.csv')

# 2. Data Cleaning: Convert string/object columns to proper numeric types
# Remove currency symbol (₹), commas (,), and percentage signs (%)
df['discounted_price'] = df['discounted_price'].str.replace('₹', '').str.replace(',', '').astype(float)

df['actual_price'] = df['actual_price'].str.replace('₹', '').str.replace(',', '').astype(float)

df['discount_percentage'] = df['discount_percentage'].str.replace('%', '').astype(float)

# Handle anomalies in 'rating' (replace anomalous '|' characters with median/mode default value like 4.0)
df['rating'] = df['rating'].str.replace('|', '4.0', regex=False).astype(float)
df['rating_count'] = df['rating_count'].str.replace(',', '').astype(float)

# 3. Generate statistical summary of all cleaned numeric columns
print("\n--- Statistical Summary of Numeric Columns ---")
numeric_cols = ['discounted_price', 'actual_price', 'discount_percentage', 'rating', 'rating_count']
print(df[numeric_cols].describe().round(2))

# 4. Target Variable Analysis (Target: 'discounted_price')
# In an e-commerce context, predicting the final listed/selling price is our primary target.
target_stats = df['discounted_price'].agg(['min', 'max', 'mean', 'median'])

print("=" * 80)

print("\n--- Target Variable ('discounted_price') Metrics ---")
print(f"Minimum Price:  ₹{target_stats['min']:.2f}")
print(f"Maximum Price:  ₹{target_stats['max']:.2f}")
print(f"Mean Price:     ₹{target_stats['mean']:.2f}")
print(f"Median Price:   ₹{target_stats['median']:.2f}")
print("=" * 90)
```

```python
# Question 5

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load the dataset
df = pd.read_csv('amazon.csv')

# Clean string columns into proper numeric types
df['discounted_price'] = df['discounted_price'].str.replace('₹', '').str.replace(',', '').astype(float)

df['actual_price'] = df['actual_price'].str.replace('₹', '').str.replace(',', '').astype(float)

df['discount_percentage'] = df['discount_percentage'].str.replace('%', '').astype(float)

df['rating'] = df['rating'].str.replace('|', '4.0', regex=False).astype(float)
df['rating_count'] = df['rating_count'].str.replace(',', '').astype(float)

# Drop rows with missing values in rating_count for clean plotting
df_clean = df.dropna(subset=['rating_count'])

# Set up the visualization style
sns.set_theme(style="whitegrid")
fig, axes = plt.subplots(3, 2, figsize=(16, 18))
axes = axes.flatten()

numeric_cols = ['discounted_price', 'actual_price', 'discount_percentage', 'rating', 'rating_count']

colors = ['#1f77b4', '#ff7f0e', '#2ca02c', '#d62728', '#9467bd']

# Plot histograms
for idx, col in enumerate(numeric_cols):
    sns.histplot(df_clean[col], kde=True, ax=axes[idx], color=colors[idx], bins=30)
    axes[idx].set_title(f'Distribution Histogram of {col.replace("_", " ").title()}', fontsize=14, pad=10)
    axes[idx].set_xlabel(col.replace('_', ' ').title(), fontsize=11)
    axes[idx].set_ylabel('Frequency Count', fontsize=11)

# Delete the 6th unused subplot axis to keep the grid clean
fig.delaxes(axes[5])

plt.tight_layout()
plt.show()
```

```python
# Question 6

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Load the dataset
df = pd.read_csv('amazon.csv')

# 2. Extract the primary top-level category for clean visualization
df['main_category'] = df['category'].str.split('|').str[0]

# Display identified categorical features suitable for analysis
print("--- Useful Categorical Feature Identified ---")
print(f"Feature: 'category' -> Extracted Top-Tier Attribute: 'main_category'")
print(f"Unique Main Categories Found: {df['main_category'].nunique()}\n")

# 3. Create the Count Plot using Seaborn
plt.figure(figsize=(12, 6))
sns.set_theme(style="whitegrid")

# Order by frequency for visual clarity
category_order = df['main_category'].value_counts().index

ax = sns.countplot(
    y='main_category',
    data=df,
    order=category_order,
    palette='viridis'
)

# Add value labels on the bars for precision
for container in ax.containers:
    ax.bar_label(container, padding=5, fontsize=10, weight='bold')

plt.title('Product Distribution across Main Categories on Amazon', fontsize=14, pad=15)
plt.xlabel('Number of Listings (Count)', fontsize=12)
plt.ylabel('Main Category / Department', fontsize=12)

plt.tight_layout()
plt.show()
```

```python
# Question 7

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Load the dataset
df = pd.read_csv('amazon.csv')

# 2. Data Cleaning: Convert object columns to clean floats
df['discounted_price'] = df['discounted_price'].str.replace('₹', '').str.replace(',', '').astype(float)
df['actual_price'] = df['actual_price'].str.replace('₹', '').str.replace(',', '').astype(float)
df['discount_percentage'] = df['discount_percentage'].str.replace('%', '').astype(float)
df['rating'] = df['rating'].str.replace('|', '4.0', regex=False).astype(float)
df['rating_count'] = df['rating_count'].str.replace(',', '').astype(float)

# Select key numeric columns
numeric_cols = ['discounted_price', 'actual_price', 'discount_percentage', 'rating', 'rating_count']
df_numeric = df[numeric_cols]

# 3. Calculate Correlation Matrix
correlation_matrix = df_numeric.corr()

# 4. Plot Heatmap
plt.figure(figsize=(10, 8))
sns.set_theme(style="white")

# Generate a mask for the upper triangle to make it cleaner (optional, but professional)
# np.triu cuts out redundant mirror images
sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap='coolwarm',
    fmt=".3f",
    linewidths=1,
    vmin=-1,
    vmax=1,
    square=True
)

plt.title('Correlation Heatmap of Amazon Product Metrics', fontsize=14, pad=15)
plt.tight_layout()
plt.show()

# 5. Display exact correlation values with the target
print("--- Correlation Matrix Values ---")
print(correlation_matrix)
```

```python
# Question 8

import pandas as pd
from sklearn.preprocessing import LabelEncoder

# Load dataset
df = pd.read_csv("amazon.csv")

# Remove Extra Spaces From Column Names
df.columns = df.columns.str.strip()

# Create A Copy For Encoding
df_encoded = df.copy()

# Encode All Categorical Columns
le = LabelEncoder()
for col in df_encoded.select_dtypes(include=['object']).columns:
    df_encoded[col] = le.fit_transform(df_encoded[col].astype(str))

# Check if Target Columns Exists
if 'discounted_price' not in df_encoded.columns:
  raise ValueError("Discounted Price Column not Found In Dataset")

# Dependent Variable
y = df_encoded['discounted_price']

# Columns To Drop From Independent Variable
columns_to_drop = [
    'product_id'
    'product_name',
    'category',
    'discounted_price',
    'about_product',
    'user_id',
    'user_name',
    'review_id',
    'review_title',
    'review_content',
    'img_link',
    'product_link'
]

# Independent Variable
X = df_encoded.drop(columns=columns_to_drop,errors='ignore')

# Display Results
print("Dependent Variable (y) Shape:",y.shape)
print("Independent Variable (X) Shape:",X.shape)

print("\n Independent Variable Columns:")
print(X.columns.tolist())

print("\n First 5 Rows Of X Variable:")
print(X.head())

print("\n First 5 Rows Of y Variable :")
print(y.head())
```

```python
import pandas as pd

# 1. Load the dataset
df = pd.read_csv('amazon.csv')

# 2. Extract two distinct categorical columns from the hierarchical 'category' column
df['main_category'] = df['category'].str.split('|').str[0]
df['sub_category'] = df['category'].str.split('|').str[1]

# Pick a subset of rows that clearly demonstrates different categories
sample_indices = [0, 1, 448]  # Targets Computers&Accessories and Home&Kitchen

df_sample = df.loc[sample_indices, ['main_category', 'sub_category']].copy()


# SHOW BEFORE TRANSFORMATION

print("=== BEFORE TRANSFORMATION (Categorical Strings) ===")
print(df_sample)
print("-" * 60)

# 3. Apply One-Hot Encoding using pandas get_dummies
# We include drop_first=True to protect the regression model from the dummy variable trap
df_encoded = pd.get_dummies(df_sample, columns=['main_category', 'sub_category'], drop_first=True)


# SHOW AFTER TRANSFORMATION

print("=== AFTER TRANSFORMATION (One-Hot Encoded Numeric Columns) ===")
# Cast boolean True/False outputs to 1/0 for clear matrix viewing
print(df_encoded.astype(int).to_string())
```
