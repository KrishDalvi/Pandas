```python
import pandas as pd
import numpy as np
import joblib
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score
```

```python
# Question 1

# Q1. Preparing Features

print("--- Q1: Preparing Features ---")

# Load your cleaned Ford Car Dataset
df = pd.read_csv('ford_cleaned.csv')

# Separate the dataset into Independent Features (X) and Dependent Feature (Y)
Y = df['price']
X = df.drop(columns=['price'])  # Drop the target column from X

# Print the shape of X and Y
print(f"Shape of X: {X.shape}")
print(f"Shape of Y: {Y.shape}\n")
```

```python
# Question 2

# Q2. One-Hot Encoding

print("--- Q2: One-Hot Encoding ---")
# Identify all categorical columns in X
categorical_cols = X.select_dtypes(include=['object']).columns.tolist()
print(f"\nCategorical columns identified: {categorical_cols}")

# Perform One-Hot Encoding using pd.get_dummies() and drop_first=True to avoid dummy variable trap
X = pd.get_dummies(X, columns=categorical_cols, drop_first=True)

# Convert the encoded data to integer type (astype(int))
X = X.astype(int)

# Show the first 5 rows after encoding
print("\nFirst 5 rows after encoding:")
print(X.head().to_string())
print("=" * 175)
```

```python
#  Question 3

# Q3. Feature Scaling

print("--- Q3: Feature Scaling ---")
# Apply Standard Scaling on the numerical columns
numerical_cols = ['year', 'mileage', 'tax', 'mpg', 'engineSize']

# Use StandardScaler from sklearn.preprocessing
scaler = StandardScaler()
X[numerical_cols] = scaler.fit_transform(X[numerical_cols])

# Show the first 5 rows of the scaled features
print("First 5 rows after feature scaling:")
print(X.head().to_string())
print("=" * 190)
```

```python
# Question 4

# Q4. Train-Test Split

print("--- Q4: Train-Test Split ---")
# Split the preprocessed data into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(X, Y, test_size=0.33, random_state=42)

# Print the shape of X_train, X_test, y_train, and y_test
print(f"Shape of X_train: {X_train.shape}")
print(f"Shape of X_test:  {X_test.shape}")
print(f"Shape of y_train: {y_train.shape}")
print(f"Shape of y_test:  {y_test.shape}\n")
print("=" * 50)
```

```python
# Question 5

# Q5. Building Linear Regression Model

print("--- Q5: Building Linear Regression Model ---")
# Train a Linear Regression model using sklearn
model = LinearRegression()

# Fit the model on the training data
model.fit(X_train, y_train)

# Print the intercept and coefficients of the model
print(f"\nIntercept: {model.intercept_:.4f}")
print("\nCoefficients:")
for col, coef in zip(X.columns, model.coef_):
    print(f"  {col}: {coef:.4f}")

print("=" * 50)
```

```python
# Question 6

# Q6. Making Predictions

print("\n--- Q6: Making Predictions ---")

# Use the trained model to make predictions on the test data and store in y_pred
y_pred = model.predict(X_test)

# Print the first 10 predicted values and compare them with the first 10 actual values from y_test
comparison_df = pd.DataFrame({
    'Actual Values (y_test)': y_test.head(10).values,
    'Predicted Values (y_pred)': y_pred[:10].round(2)
})

print(comparison_df.to_string())
print("\n")
print("=" * 60)
```

```python
# Question 7


# Q7. Model Evaluation using R² Score

print("\n--- Q7: Model Evaluation using R² Score ---")

# Calculate the R² score
r2 = r2_score(y_test, y_pred)

print(f"\nR² Score: {r2:.4f}\n")
print("=" * 50)
```

```python
# Question 8

# Q8. Saving the Model

print("\n--- Q8: Saving the Model ---")

# Save your trained model and preprocessing objects using joblib
joblib.dump(model, 'LR_ford_car.pkl')
joblib.dump(scaler, 'scaler.pkl')
joblib.dump(X.columns.tolist(), 'columns.pkl')

print("\nSaved files successfully:")
print("  - LR_ford_car.pkl\n  - scaler.pkl\n  - columns.pkl\n")
print("=" * 40)
```
