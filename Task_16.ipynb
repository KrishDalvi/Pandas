```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load dataset
df = pd.read_csv("insurance.csv")

# Remove extra spaces from column names
df.columns = df.columns.str.strip()

# Rename 'expenses' to 'charges' if needed
if 'expenses' in df.columns:
    df.rename(columns={'expenses': 'charges'}, inplace=True)

# Check column names
print(df.columns)

# Distribution plots
# Age
plt.figure(figsize=(6,4))
sns.histplot(df['age'], kde=True, color='cornflowerblue')
plt.title("Distribution of Age")
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.show()

# BMI
plt.figure(figsize=(6,4))
sns.histplot(df['bmi'], kde=True, color='cornflowerblue')
plt.title("Distribution of BMI")
plt.xlabel("BMI")
plt.ylabel("Frequency")
plt.show()

# Children
plt.figure(figsize=(6,4))
sns.histplot(df['children'], discrete=True, color='cornflowerblue')
plt.title("Distribution of Children")
plt.xlabel("Number of Children")
plt.ylabel("Frequency")
plt.show()

# Charges
plt.figure(figsize=(6,4))
sns.histplot(df['charges'], kde=True, color='cornflowerblue')
plt.title("Distribution of Charges")
plt.xlabel("Charges")
plt.ylabel("Frequency")
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Define categorical columns
categorical_cols = ['sex', 'smoker', 'region']

# Q6. Count plots for all categorical columns
for col in categorical_cols:
    plt.figure(figsize=(6, 4))

    # Sort bars from highest frequency to lowest
    order = df[col].value_counts().index

    sns.countplot(
        data=df,
        x=col,
        order=order,
        hue=col,          # Required to avoid future seaborn warning
        palette='muted',
        legend=False
    )

    plt.title(f'Count Plot of {col.capitalize()}', fontsize=12, fontweight='bold')
    plt.xlabel(col.capitalize())
    plt.ylabel('Count')
    plt.tight_layout()
    plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
# Q 7
# Define numerical columns
numerical_cols = ['age', 'bmi', 'children', 'charges']

# If your dataset has 'expenses' instead of 'charges'
if 'expenses' in df.columns:
    df.rename(columns={'expenses': 'charges'}, inplace=True)

# Correlation Heatmap
plt.figure(figsize=(6, 5))

sns.heatmap(
    df[numerical_cols].corr(),
    annot=True,
    cmap='YlGnBu',
    fmt='.2f',
    linewidths=0.5
)

plt.title('Correlation Heatmap', fontsize=13, fontweight='bold')
plt.tight_layout()
plt.show()
```

```python
# Q8. Statistical Analysis on the Charges column
print("--- Q8: Charges Data Analysis ---")
print(f"\nAverage Insurance Charges: ${df['charges'].mean():.2f}")
print(f"\nMaximum Insurance Charges: ${df['charges'].max():.2f}")
print(f"\nMinimum Insurance Charges: ${df['charges'].min():.2f}\n")

print("\nAverage Charges Grouped by Smoker Status:")
print(df.groupby('smoker')['charges'].mean().round(2))
print("\n" + "="*60 + "\n")
```

```python
# ------------------
# Q9. Boxplots
# Compare charges based on smoker and sex
# ------------------

plt.figure(figsize=(6,4))
sns.boxplot(x="smoker", y="charges", data=df)
plt.title("Charges by Smoker")
plt.show()

plt.figure(figsize=(6,4))
sns.boxplot(x="sex", y="charges", data=df)
plt.title("Charges by Sex")
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Remove extra spaces from column names
df.columns = df.columns.str.strip()

# Rename 'expenses' to 'charges' if needed
if 'expenses' in df.columns:
    df.rename(columns={'expenses': 'charges'}, inplace=True)

# Boxplot of Charges by Smoker and Sex
plt.figure(figsize=(7, 5))

sns.boxplot(
    data=df,
    x='smoker',
    y='charges',
    hue='sex',
    palette='muted'
)

plt.title('Charges Distribution by Smoker Status and Sex', fontsize=13, fontweight='bold')
plt.xlabel('Smoker Status')
plt.ylabel('Insurance Charges ($)')
plt.tight_layout()
plt.show()
```
